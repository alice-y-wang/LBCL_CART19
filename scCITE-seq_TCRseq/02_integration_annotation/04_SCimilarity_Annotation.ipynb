```python
# load packages
from scimilarity.utils import lognorm_counts, align_dataset
from scimilarity import CellQuery
from scimilarity import CellAnnotation
import numpy as np
import pandas as pd
import scanpy as sc
from matplotlib import pyplot as plt
```

```python
# set parameters
sc.set_figure_params(dpi=100)
plt.rcParams["figure.figsize"] = [6, 4]

import warnings

warnings.filterwarnings("ignore")
```

```python
model_annotation_path = "./scimilarity/models/model_v1.1"
ca = CellAnnotation(model_path=model_annotation_path)
```

```python
ca
```

```python
# load Adata object
adata_tnk = sc.read_h5ad("./seurat_objects/all_samples_TNKonly.h5ad")
adata_mono = sc.read_h5ad("./seurat_objects/all_samples_MONOonly.h5ad")
```

```python
# save raw adata as new tmp adata file
adata_tnk_raw = adata_tnk.raw.to_adata()
adata_mono_raw = adata_mono.raw.to_adata()
```

```python
# do adata tnk counts 
adata_tnk.layers["counts"] = adata_tnk_raw.X.copy()
adata_mono.layers["counts"] = adata_mono_raw.X.copy()
```

```python
adata_tnk.obs = adata_tnk.obs[[ 'orig.ident', 'nCount_RNA', 'nFeature_RNA', 'nCount_ADT', 'nFeature_ADT', 
                       'percent.mt', 'percent.ribo', 'barcode', 'raw_clonotype_id', 
                       'chain', 'v_gene', 'j_gene', 'd_gene', 'c_gene', 'productive', 
                       'seurat_clusters', 'S.Score', 'G2M.Score', 'Phase', 'HeatShock.Score1', 
                        'Doublet_Singlet', 'predicted.celltype.l1.score', 'predicted.celltype.l1', 
                       'predicted.celltype.l2.score', 'predicted.celltype.l2', 'predicted.celltype.l3.score', 
                       'predicted.celltype.l3', 'sample.name','patient.id', 'response', 'str_split', 
                       'timepoint', 'timepoint_order','harmony.snn_res.0.6', 'harmony.snn_res.0.8', 
                       'harmony.snn_res.1', 'harmony.snn_res.1.5', 'harmony.snn_res.2', 'harmony.snn_res.0.2',
                       'harmony.snn_res.0.4', 'harmonyumap_1', 'harmonyumap_2', 
                       'basic.anno', 'int.snn_res.0.6', 'int.snn_res.0.8', 
                       'int.snn_res.1', 'int.snn_res.1.5', 'int.snn_res.2']]

adata_tnk.var = adata_tnk.var[['var.features', 'var.features.rank']]
```

```python
adata_mono.obs = adata_mono.obs[[ 'orig.ident', 'nCount_RNA', 'nFeature_RNA', 'nCount_ADT', 'nFeature_ADT', 
                       'percent.mt', 'percent.ribo', 'barcode', 'raw_clonotype_id', 
                       'chain', 'v_gene', 'j_gene', 'd_gene', 'c_gene', 'productive', 
                       'seurat_clusters', 'S.Score', 'G2M.Score', 'Phase', 'HeatShock.Score1', 
                        'Doublet_Singlet', 'predicted.celltype.l1.score', 'predicted.celltype.l1', 
                       'predicted.celltype.l2.score', 'predicted.celltype.l2', 'predicted.celltype.l3.score', 
                       'predicted.celltype.l3', 'sample.name','patient.id', 'response', 'str_split', 
                       'timepoint', 'timepoint_order','harmony.snn_res.0.6', 'harmony.snn_res.0.8', 
                       'harmony.snn_res.1', 'harmony.snn_res.1.5', 'harmony.snn_res.2', 'harmony.snn_res.0.2',
                       'harmony.snn_res.0.4', 'harmonyumap_1', 'harmonyumap_2', 
                       'basic.anno', 'int.snn_res.0.6', 'int.snn_res.0.8', 
                       'int.snn_res.1', 'int.snn_res.1.5', 'int.snn_res.2']]

adata_mono.var = adata_mono.var[['var.features', 'var.features.rank']]
```

```python
# match feature space with SCimiliarity model
adata_tnk = align_dataset(adata_tnk, ca.gene_order)
adata_mono = align_dataset(adata_mono, ca.gene_order)
```

```python
# check 
adata_tnk.layers
adata_mono.layers
```

```python
# normalize data to match with SCimilarity
adata_tnk = lognorm_counts(adata_tnk)
adata_mono= lognorm_counts(adata_mono)
```

# set a safelist

```python
mono_target_celltypes = [
    "CD141-positive myeloid dendritic cell",
    "CD1c-positive myeloid dendritic cell",
    "classical monocyte",
    "conventional dendritic cell",
    "intermediate monocyte",
    "non-classical monocyte",
    "plasmacytoid dendritic cell, human"
]
```

```python
tnk_target_celltypes = [
    "CD16-negative, CD56-bright natural killer cell, human",
    "CD16-positive, CD56-dim natural killer cell, human",
    "CD4-positive, alpha-beta cytotoxic T cell",
    "CD8-positive, alpha-beta cytotoxic T cell",
    "T-helper 1 cell",
    "T-helper 17 cell",
    "activated CD4-positive, alpha-beta T cell",
    "activated CD8-positive, alpha-beta T cell",
    "central memory CD4-positive, alpha-beta T cell",
    "central memory CD8-positive, alpha-beta T cell",
    "effector CD4-positive, alpha-beta T cell",
    "effector CD8-positive, alpha-beta T cell",
    "effector memory CD4-positive, alpha-beta T cell",
    "effector memory CD8-positive, alpha-beta T cell",
    "effector memory CD8-positive, alpha-beta T cell, terminally differentiated",
    "gamma-delta T cell",
    "innate lymphoid cell",
    "mature NK T cell",
    "mucosal invariant T cell",
    "naive T cell",
    "naive regulatory T cell",
    "natural killer cell",
    "regulatory T cell"
]
```

```python
ca.safelist_celltypes(mono_target_celltypes)
ca.safelist_celltypes(tnk_target_celltypes)
```

```python
adata_mono = ca.annotate_dataset(adata_mono)
adata_tnk = ca.annotate_dataset(adata_tnk)
```

```python
# get embeddings
adata_tnk.obsm["X_scimilarity"] = ca.get_embeddings(adata_tnk.X)
adata_mono.obsm["X_scimilarity"] = ca.get_embeddings(adata_mono.X)
```

```python
# Visualize
sc.pp.neighbors(adata_tnk, use_rep="X_scimilarity")
sc.tl.umap(adata_tnk)

sc.pp.neighbors(adata_mono, use_rep="X_scimilarity")
sc.tl.umap(adata_mono)

sc.pl.umap(adata_tnk, color="celltype_hint", legend_fontsize=5)

sc.pl.umap(adata_tnk, color="predicted.celltype.l2", legend_fontsize=5)

sc.pl.umap(adata_tnk, color = "predicted.celltype.l2")
sc.pl.umap(adata_mono, color = "predicted.celltype.l2")
```

# Unconstrained Annotation

```python
# unconstrained annotations TNK
predictions, nn_idxs, nn_dists, nn_stats = ca.get_predictions_knn(
   adata_tnk.obsm["X_scimilarity"]
)
adata_tnk.obs["predictions_unconstrained"] = predictions.values

adata_tnk.obs["celltype_hint"] = predictions.values
adata_tnk.obs["min_dist"] = nn_stats["min_dist"].values
adata_tnk.obs["celltype_hits"] = nn_stats["hits"].values
adata_tnk.obs["celltype_hits_weighted"] = nn_stats["hits_weighted"].values
adata_tnk.obs["celltype_hint_stat"] = nn_stats["vsAll"].values
adata_tnk.obs["celltype_hint_weighted_stat"] = nn_stats["vsAll_weighted"].values
```

```python
# unconstrained annotations MONO
predictions, nn_idxs, nn_dists, nn_stats = ca.get_predictions_knn(
    adata_mono.obsm["X_scimilarity"]
)
adata_mono.obs["predictions_unconstrained"] = predictions.values

adata_mono.obs["celltype_hint"] = predictions.values
adata_mono.obs["min_dist"] = nn_stats["min_dist"].values
adata_mono.obs["celltype_hits"] = nn_stats["hits"].values
adata_mono.obs["celltype_hits_weighted"] = nn_stats["hits_weighted"].values
adata_mono.obs["celltype_hint_stat"] = nn_stats["vsAll"].values
adata_mono.obs["celltype_hint_weighted_stat"] = nn_stats["vsAll_weighted"].values
```

```python
celltype_counts = adata_tnk.obs.predictions_unconstrained.value_counts()
well_represented_celltypes = celltype_counts[celltype_counts > 20].index

celltype_counts = adata_mono.obs.predictions_unconstrained.value_counts()
well_represented_celltypes = celltype_counts[celltype_counts > 20].index
```
